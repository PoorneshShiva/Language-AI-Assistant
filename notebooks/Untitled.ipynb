```python
import wave
import sys

import pyaudio

CHUNK = 1024
FORMAT = pyaudio.paInt16
CHANNELS = 1 if sys.platform == 'darwin' else 2
RATE = 44100
RECORD_SECONDS = 5

with wave.open('output.mp3', 'wb') as wf:
    p = pyaudio.PyAudio()
    wf.setnchannels(CHANNELS)
    wf.setsampwidth(p.get_sample_size(FORMAT))
    wf.setframerate(RATE)

    stream = p.open(format=FORMAT, channels=CHANNELS, rate=RATE, input=True)

    print('Recording...')
    for _ in range(0, RATE // CHUNK * RECORD_SECONDS):
        wf.writeframes(stream.read(CHUNK))
    print('Done')

    stream.close()
    p.terminate()
```

```python
from langchain.text_splitter import TokenTextSplitter
for index, path in enumerate(all_data_path):
    loader = TextLoader(path)
    docs = loader.load()
    # text_splitter = RecursiveCharacterTextSplitter(chunk_size=80,
    #                                                chunk_overlap=50)
    # documents = text_splitter.split_documents(docs)

    text_splitter = TokenTextSplitter(chunk_size=50, chunk_overlap=30)
    
    documents = []
    for doc in docs:
        log_entries = doc.page_content.split('\n')
        for entry in log_entries:
            split_docs = text_splitter.split_text(entry)
            for split_doc in split_docs:
                documents.append(Document(page_content=split_doc, metadata={"id": index}))
    

    vector_store.add_documents(documents, id=index)
```

```python
from langchain_community.llms import Ollama
from langchain.chains import RetrievalQA
from langchain.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnableMap

# Assuming you have a vector store (like PGVector, FAISS, or Chroma)
retriever = vector_store.as_retriever(search_type="similarity", search_kwargs={"k": 3})

# Load LLM
llm = Ollama(model="llama3.2")

# Define a Prompt Template
template = """You are an AI assistant answering questions based on retrieved documents.

Context:
{context}

Question:
{question}

Answer:"""

prompt = ChatPromptTemplate.from_template(template)

# RAG Chain with Map-Rerank
rag_chain = RetrievalQA.from_chain_type(
    llm=llm,
    retriever=retriever,
    chain_type="map_rerank",
)

query = "When was a Critical Radiation Alert?"
response = rag_chain.run(query)

print("RAG Response:", response)

# Combining RAG with a Prompt Template
rag_pipeline = RunnableMap({
    "context": retriever,
    "question": lambda x: x["question"]
}) | prompt | llm

response = rag_pipeline.invoke({"question": "What are the usage of a large language model?"})
print("Final Response:", response)
```
